# **Objective**


The objective of this project is to analyze vendor performance by examining sales, purchasing, profitability, freight costs, and inventory efficiency. The analysis aims to identify high- and low-performing vendors, understand the factors affecting profitability and operational efficiency, and provide data-driven insights to support better vendor selection, cost reduction, inventory management, and overall business performance.

# **Problem Statement**

1. Analyze vendor sales, purchases, and overall contribution to the business.
2. Identify high-performing and underperforming vendors based on sales and profitability.
3. Evaluate profits and identify the most and least profitable products.
4. Analyze purchase, freight, and other costs that affect profitability.
5. Compare sales and purchase values to measure vendor performance.
6. Evaluate inventory efficiency and identify fast- and slow-moving products.
7. Examine pricing differences and their impact on revenue and profit.
8. Compare vendors and brands to identify areas for improvement.
9. Identify the vendors and products that are making losses.
10. Evaluate how stock turnover and order size relates to sales and profit margin.


# **Outcome**

The analysis will identify the best- and worst-performing vendors and products based on sales, profit, and profit margins. It will highlight high purchase and freight costs, slow-moving inventory, and other factors affecting profitability. The analysis will also reveal which vendors and brands provide the best overall value to the business. Finally, the findings will help recommend ways to reduce costs, improve inventory management, and increase overall profitability.

| Column Name | Data Type | Description |
|---|---|---|
| `VendorNumber` | Integer | Unique identification number assigned to each vendor. |

| `Brand` | Integer | Unique identification number of the product brand. |

| `PurchasePrice` | Float | Purchase price paid by the business for one unit of the product. |

| `TotalPurchaseQuantity` | Integer | Total number of units purchased from the vendor. |

| `TotalExciseTax` | Float | Total excise tax associated with the purchased or sold products. |
| `FreightCost` | Float | Total transportation or freight cost incurred for receiving the products. |
| `GrossProfit` | Float | Total profit generated after accounting for the purchase cost and related costs from sales revenue. |
| `ProfitMargin` | Float | Percentage of sales revenue retained as profit after relevant costs. |
| `StockTurnover` | Float | Measure of how efficiently inventory is sold or replaced during the period. |
| `SalesToPurchaseRatio` | Float | Ratio comparing total sales value with total purchase value, indicating sales performance relative to purchasing cost. |

**1. What are the total purchases, sales, gross profit and profit margin?**

```python
total_purchase = df["TotalPurchaseDollars"].sum()
total_sales = df["TotalSalesDollars"].sum()
total_profit = df["GrossProfit"].sum()
overall_margin = (total_profit / total_sales * 100)


print(f"Total Purchase Value: ${total_purchase:,.2f}")
print(f"Total Sales: ${total_sales:,.2f}")
print(f"Total Gross Profit: ${total_profit:,.2f}")
print(f"Total Profit Margin: {overall_margin:.2f}%")
```

```python
overall_performance=df[['TotalPurchaseDollars',"TotalSalesDollars","GrossProfit"]].sum()
overall_performance["ProfitMargin"] = (
    overall_performance["GrossProfit"]
    / overall_performance["TotalSalesDollars"]
    * 100
)

display(overall_performance)
```

```python
plt.figure(figsize=(10,6))
bars=plt.bar(overall_performance.index,overall_performance.values)
plt.bar_label(bars, fmt='{:,.2f}', padding=3, fontsize=9)

plt.title("Overall Business Financial Performance")
plt.show()
```

```python
# Calculate overall profit margin using df

total_sales = df["TotalSalesDollars"].sum()

total_profit = df.loc[
    df["GrossProfit"] > 0, "GrossProfit"
].sum()

total_loss = df.loc[
    df["GrossProfit"] < 0, "GrossProfit"
].sum()

net_profit_loss = total_profit + total_loss

positive_profit_margin = (total_profit / total_sales) * 100
loss_margin = (total_loss / total_sales) * 100
net_profit_margin = (net_profit_loss / total_sales) * 100

print(f"Positive Profit Margin: {positive_profit_margin:.2f}%")
print(f"Loss Margin: {loss_margin:.2f}%")
print(f"Overall Net Profit Margin: {net_profit_margin:.2f}%")
```

## **Hypothesis Testing (T-Test)**

**12. Do profitable and loss-making transactions differ in their gross profit values?**

```python
from scipy.stats import ttest_ind

profit = df[df['GrossProfit'] > 0]['GrossProfit']
loss = df[df['GrossProfit'] < 0]['GrossProfit']

print("Profit:")
print(profit)

print("Loss:")
print(loss)
```

```python
ttest_ind(profit, loss)
```

**Insight:**

The p-value is far below 0.05, indicating a statistically **significant difference** in **gross profit** between **profitable** and **loss-making** transactions.

**13. Does average gross profit differ significantly across the four freight cost categories?**

**Insight:**
